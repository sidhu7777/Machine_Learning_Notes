````python
from sklearn.datasets import load_digits
img=load_digits().images[0]; kernel=np.array([[1,0,-1],[1,0,-1],[1,0,-1]]); out=np.zeros((6,6))
for i in range(6):
 for j in range(6): out[i,j]=np.sum(img[i:i+3,j:j+3]*kernel)
fig,ax=plt.subplots(1,3,figsize=(10,3)); ax[0].imshow(img,cmap='gray'); ax[1].imshow(kernel,cmap='coolwarm'); ax[2].imshow(out,cmap='gray'); [a.axis('off') for a in ax]; plt.show()
````

````python
pooled=np.array([[out[i:i+2,j:j+2].max() for j in range(0,5,2)] for i in range(0,5,2)]); print(out.shape,pooled.shape); display(pd.DataFrame(pooled).round(2))
````
