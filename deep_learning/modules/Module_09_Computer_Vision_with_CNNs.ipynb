## Research-Backed Coverage Expansion: Computer Vision with CNNs

This expansion fills concepts commonly covered in deep-learning texts and courses such as the Deep Learning book, CS231n/CS224n, and Dive into Deep Learning. Each concept is written for learning flow: definition, where it is used, why it matters, and common mistakes.


### Channel Ordering

**Definition:** Images may be stored as HWC or CHW.

**Where it is used:** NumPy/PIL often use HWC; PyTorch commonly uses CHW.

**Why it matters:** Wrong ordering breaks convolution input shapes.

**Common mistake:** Treating height as channel dimension.


### Kernel as Pattern Detector

**Definition:** A convolution kernel responds strongly to matching local patterns.

**Where it is used:** Edges, textures, corners, and learned CNN filters.

**Why it matters:** It explains why CNNs learn spatial features.

**Common mistake:** Thinking kernels are manually chosen in modern CNNs.


### Cross-Correlation vs Mathematical Convolution

**Definition:** Deep-learning libraries usually implement cross-correlation, not flipped-kernel convolution.

**Where it is used:** CNN layers.

**Why it matters:** The learned filter adapts, so naming difference rarely matters in practice.

**Common mistake:** Manually flipping kernels when the framework already defines behavior.


### Output Size Formula

**Definition:** For input size n, kernel k, padding p, stride s: output floor((n+2p-k)/s)+1.

**Where it is used:** CNN shape debugging.

**Why it matters:** Prevents layer-size surprises.

**Common mistake:** Ignoring floor behavior with stride.


### Dilation

**Definition:** Spacing between kernel elements.

**Where it is used:** Segmentation and receptive-field expansion.

**Why it matters:** It increases receptive field without increasing parameters much.

**Common mistake:** Using dilation without checking gridding artifacts.


### Transposed Convolution

**Definition:** A learned upsampling operation.

**Where it is used:** Segmentation, autoencoders, GANs.

**Why it matters:** It increases spatial resolution.

**Common mistake:** Calling it exact inverse convolution.


### Global Average Pooling

**Definition:** Averages each feature map into one number.

**Where it is used:** Modern CNN classifiers.

**Why it matters:** It reduces parameters compared with large dense heads.

**Common mistake:** Flattening huge feature maps into massive dense layers.


### Translation Equivariance vs Invariance

**Definition:** Convolution is equivariant; pooling/aggregation can create partial invariance.

**Where it is used:** Vision architecture reasoning.

**Why it matters:** A shifted object shifts feature maps, while classification should often stay same.

**Common mistake:** Saying convolution alone is fully invariant.


### Data Augmentation for Vision

**Definition:** Random crop, flip, color jitter, Cutout, MixUp, CutMix.

**Where it is used:** Vision training and robustness.

**Why it matters:** It teaches invariances and reduces overfitting.

**Common mistake:** Using vertical flip for labels where vertical orientation matters.


### Segmentation vs Classification

**Definition:** Classification predicts one label for image; segmentation predicts labels per pixel.

**Where it is used:** Medical imaging, autonomous driving, remote sensing.

**Why it matters:** Output shape and loss are different.

**Common mistake:** Using image-level labels for pixel-level evaluation.


### Object Detection Basics

**Definition:** Predicts boxes, classes, and confidence.

**Where it is used:** YOLO/Faster R-CNN-style tasks.

**Why it matters:** Adds localization to classification.

**Common mistake:** Evaluating detection with plain accuracy.


### CNN Interpretability Preview

**Definition:** Filters, activation maps, and Grad-CAM show where features respond.

**Where it is used:** Debugging and trust.

**Why it matters:** Helps find shortcut learning.

**Common mistake:** Assuming highlighted region is always causal proof.


```python
import numpy as np, matplotlib.pyplot as plt
img = np.zeros((8,8)); img[2:6, 3:5] = 1
kernel = np.array([[1,0,-1],[1,0,-1],[1,0,-1]])
out = np.zeros((6,6))
for i in range(6):
    for j in range(6):
        out[i,j] = np.sum(img[i:i+3,j:j+3]*kernel)
fig, ax = plt.subplots(1,3,figsize=(10,3))
ax[0].imshow(img,cmap='gray'); ax[0].set_title('image')
ax[1].imshow(kernel,cmap='coolwarm'); ax[1].set_title('vertical-edge kernel')
ax[2].imshow(out,cmap='coolwarm'); ax[2].set_title('feature map response')
for a in ax: a.axis('off')
```

## Algorithm and Dataset Expansion: CNN Algorithms and Vision Datasets

This section connects the concepts to algorithms and datasets. For each module, the learner should understand what data shape is appropriate, which algorithm family fits the data, what metric to read, and what failure pattern to check.


### CNN Algorithm Components

| Component | What algorithm does | Dataset example |
|---|---|---|
| Convolution | scans local kernels across image | MNIST/digits edges and strokes |
| Pooling | reduces spatial size | object classification |
| Strided convolution | learned downsampling | modern CNN stages |
| Dilated convolution | expands receptive field | segmentation |
| Transposed convolution | learned upsampling | autoencoders/segmentation |
| Global average pooling | converts feature maps to class vector | modern classifiers |
| Data augmentation | creates transformed training views | natural images |


### Dataset Explanation

Start with **digits/MNIST-like data** because learners can see strokes and edges. Then move to natural images like CIFAR-style data because color channels, augmentation, and object variation matter.

For CNNs, always explain tensor shape:

* grayscale image: `(height, width)` or `(1, height, width)`;
* RGB image: `(3, height, width)` in PyTorch style;
* batch of RGB images: `(batch, 3, height, width)`.


```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits=load_digits()
fig,axes=plt.subplots(2,5,figsize=(8,3))
for ax,img,label in zip(axes.ravel(),digits.images[:10],digits.target[:10]):
    ax.imshow(img,cmap='gray'); ax.set_title(f'label {label}'); ax.axis('off')
plt.suptitle('Digits dataset: good first dataset for CNN concepts');
print('images shape:', digits.images.shape, 'flattened data shape:', digits.data.shape)
```

**Metric meaning:** for balanced digit classification, accuracy is understandable. For medical images or defect detection, accuracy may be misleading and recall/precision/AUC may matter more.
