````python
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
X,y=load_digits(return_X_y=True); X=X/16; pca=PCA(n_components=2,random_state=42); Z=pca.fit_transform(X); Xr=pca.inverse_transform(Z); print(X.shape,Z.shape,Xr.shape); plt.scatter(Z[:,0],Z[:,1],c=y,cmap='tab10',s=10); plt.colorbar(); plt.title('2D latent space'); plt.show()
````

````python
fig,ax=plt.subplots(2,8,figsize=(9,2.5))
for i in range(8): ax[0,i].imshow(X[i].reshape(8,8),cmap='gray'); ax[0,i].axis('off'); ax[1,i].imshow(Xr[i].reshape(8,8),cmap='gray'); ax[1,i].axis('off')
plt.show()
````
