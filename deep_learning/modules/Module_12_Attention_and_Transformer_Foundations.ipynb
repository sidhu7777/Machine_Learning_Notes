## Research-Backed Coverage Expansion: Attention and Transformer Foundations

This expansion fills concepts commonly covered in deep-learning texts and courses such as the Deep Learning book, CS231n/CS224n, and Dive into Deep Learning. Each concept is written for learning flow: definition, where it is used, why it matters, and common mistakes.


### Attention Score

**Definition:** Similarity between a query and keys.

**Where it is used:** Self-attention and cross-attention.

**Why it matters:** Determines which tokens influence each other.

**Common mistake:** Treating attention weights as guaranteed explanation.


### Scaled Dot Product

**Definition:** Dot product divided by sqrt(dk).

**Where it is used:** Transformer attention.

**Why it matters:** Scaling prevents softmax from becoming too sharp for large dimensions.

**Common mistake:** Forgetting scale and causing saturated attention.


### Attention Matrix

**Definition:** Matrix of token-to-token attention weights.

**Where it is used:** Transformer visualization/debugging.

**Why it matters:** Shows information flow between positions.

**Common mistake:** Ignoring masks when reading attention.


### Multi-Head Diversity

**Definition:** Different heads can learn different relation types.

**Where it is used:** Transformers.

**Why it matters:** Parallel subspaces improve representation capacity.

**Common mistake:** Assuming every head is interpretable or useful.


### Position Information

**Definition:** Transformers need position encodings because attention itself is order-agnostic.

**Where it is used:** Text, audio, time series.

**Why it matters:** Order changes meaning.

**Common mistake:** Removing positions and expecting sequence understanding.


### Absolute vs Relative Position

**Definition:** Absolute encodes each location; relative encodes distance/relationship between positions.

**Where it is used:** Transformer variants.

**Why it matters:** Relative positions often generalize better to different lengths.

**Common mistake:** Using fixed absolute positions beyond trained length carelessly.


### Encoder-Only Transformer

**Definition:** Uses bidirectional attention for representation learning.

**Where it is used:** BERT-style classification/embedding models.

**Why it matters:** Good when whole input is visible.

**Common mistake:** Using encoder-only architecture for causal generation.


### Decoder-Only Transformer

**Definition:** Uses causal masked attention for next-token prediction.

**Where it is used:** GPT-style language models.

**Why it matters:** Good for generation.

**Common mistake:** Forgetting causal mask and leaking future tokens.


### Encoder-Decoder Transformer

**Definition:** Encoder reads source; decoder generates target with cross-attention.

**Where it is used:** Translation/summarization.

**Why it matters:** Good for conditional generation.

**Common mistake:** Using only self-attention when output must condition on source.


### Feed-Forward Network

**Definition:** Per-token MLP inside each transformer block.

**Where it is used:** Transformer blocks.

**Why it matters:** Adds nonlinear transformation after attention mixing.

**Common mistake:** Thinking attention is the only computation in transformers.


### Residual and LayerNorm

**Definition:** Stabilize deep transformer blocks.

**Where it is used:** All modern transformers.

**Why it matters:** Improve gradient flow and training stability.

**Common mistake:** Removing them and expecting deep transformers to train.


### Attention Complexity

**Definition:** Full attention is O(n²) in sequence length.

**Where it is used:** Long-context modeling.

**Why it matters:** Long sequences become expensive.

**Common mistake:** Increasing context length without memory planning.


```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(0)
Q=rng.normal(size=(5,4)); K=rng.normal(size=(5,4)); V=rng.normal(size=(5,3))
scores=Q@K.T/np.sqrt(K.shape[1])
weights=np.exp(scores-scores.max(axis=1,keepdims=True)); weights=weights/weights.sum(axis=1,keepdims=True)
out=weights@V
plt.imshow(weights,cmap='viridis')
plt.title('Attention matrix: each row sums to 1')
plt.xlabel('key position'); plt.ylabel('query position'); plt.colorbar();
print('attention output shape:', out.shape)
```
