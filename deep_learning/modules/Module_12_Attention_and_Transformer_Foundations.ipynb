````python
tokens=['deep','learning','uses','attention']; d=4; X=rng.normal(size=(len(tokens),d)); Wq,Wk,Wv=rng.normal(size=(d,d)),rng.normal(size=(d,d)),rng.normal(size=(d,d)); Q,K,V=X@Wq,X@Wk,X@Wv; scores=Q@K.T/np.sqrt(d); weights=np.exp(scores-scores.max(axis=1,keepdims=True)); weights/=weights.sum(axis=1,keepdims=True); display(pd.DataFrame(weights,index=tokens,columns=tokens).round(3)); plt.imshow(weights,cmap='viridis'); plt.xticks(range(4),tokens); plt.yticks(range(4),tokens); plt.colorbar(); plt.title('attention weights'); plt.show()
````

````python
n=6; mask=np.triu(np.ones((n,n)),k=1); plt.imshow(mask,cmap='gray_r'); plt.title('causal mask'); plt.show()
````
