```python
n=4; V=np.zeros((n,n)); goal=(3,3); gamma=.9
for _ in range(30):
 Vn=V.copy()
 for i in range(n):
  for j in range(n):
   if (i,j)==goal: Vn[i,j]=1; continue
   Vn[i,j]=max([-0.01+gamma*V[np.clip(i+di,0,n-1),np.clip(j+dj,0,n-1)] for di,dj in [(1,0),(-1,0),(0,1),(0,-1)]])
 V=Vn
plt.imshow(V,cmap='viridis'); plt.colorbar(); plt.title('value function'); plt.show(); display(pd.DataFrame(V).round(2))
```

## Research-Backed Coverage Expansion: Deep Reinforcement Learning

This expansion fills concepts commonly covered in deep-learning texts and courses such as the Deep Learning book, CS231n/CS224n, and Dive into Deep Learning. Each concept is written for learning flow: definition, where it is used, why it matters, and common mistakes.


### MDP

**Definition:** Markov decision process with states, actions, rewards, transitions, and discount.

**Where it is used:** RL problem formulation.

**Why it matters:** Defines what the agent is solving.

**Common mistake:** Ignoring whether the state is truly Markov.


### Exploration vs Exploitation

**Definition:** Choosing between trying new actions and using known good actions.

**Where it is used:** All RL algorithms.

**Why it matters:** Poor exploration prevents learning good policies.

**Common mistake:** Evaluating exploratory behavior as final policy.


### Bellman Equation

**Definition:** Recursive relationship between value and future value.

**Where it is used:** Value methods, Q-learning, DQN.

**Why it matters:** Core of bootstrapped RL learning.

**Common mistake:** Treating targets as fixed labels.


### Temporal-Difference Error

**Definition:** Difference between current estimate and bootstrapped target.

**Where it is used:** Q-learning, actor-critic.

**Why it matters:** Training signal for value estimates.

**Common mistake:** Ignoring high variance/instability of bootstrapping.


### Off-Policy Learning

**Definition:** Learns target policy from data generated by another policy.

**Where it is used:** Q-learning, DQN, replay buffers.

**Why it matters:** Allows reuse of old experience.

**Common mistake:** Using off-policy data with algorithms that assume on-policy.


### On-Policy Learning

**Definition:** Learns from data generated by current policy.

**Where it is used:** Policy gradient, PPO.

**Why it matters:** Stable objective assumptions.

**Common mistake:** Reusing stale data too many times.


### Replay Buffer Sampling

**Definition:** Stores and samples previous transitions.

**Where it is used:** DQN-style algorithms.

**Why it matters:** Breaks correlation and improves sample efficiency.

**Common mistake:** Forgetting distribution changes as policy improves.


### Target Network Lag

**Definition:** Slowly updated copy used for stable targets.

**Where it is used:** DQN and variants.

**Why it matters:** Reduces moving-target instability.

**Common mistake:** Updating target network every step exactly like online network.


### Policy Entropy

**Definition:** Measures randomness of action distribution.

**Where it is used:** Policy-gradient exploration.

**Why it matters:** Entropy bonus encourages exploration.

**Common mistake:** Leaving high entropy at deployment when deterministic behavior is needed.


### Advantage Function

**Definition:** How much better an action is than baseline value.

**Where it is used:** Actor-critic and PPO.

**Why it matters:** Reduces variance in policy updates.

**Common mistake:** Confusing advantage with raw reward.


### Reward Shaping

**Definition:** Adding helpful reward signals.

**Where it is used:** Sparse-reward tasks.

**Why it matters:** Can speed learning but may change behavior.

**Common mistake:** Creating rewards that teach shortcuts.


### Offline RL Risk

**Definition:** Learning from fixed logged data without environment interaction.

**Where it is used:** Healthcare, robotics logs, recommender logs.

**Why it matters:** Distribution shift can make unseen actions dangerous.

**Common mistake:** Evaluating only with ordinary supervised metrics.


```python
import numpy as np, matplotlib.pyplot as plt
gamma=0.9
rewards=[1,1,1,1,1]
returns=[sum((gamma**k)*r for k,r in enumerate(rewards[t:])) for t in range(len(rewards))]
plt.bar(range(len(returns)), returns)
plt.title('Discounted return from each time step')
plt.xlabel('time step'); plt.ylabel('return');
print('returns:', np.round(returns,3))
```

## Algorithm and Dataset Expansion: Deep RL Algorithms and Environments

### Deep RL Algorithm Map

| Algorithm | Type | Action space | Key machinery |
|---|---|---|---|
| DQN | value-based, off-policy | discrete | replay buffer, target network |
| Double DQN | value-based | discrete | reduces max overestimation |
| Dueling DQN | value-based | discrete | separates value and advantage |
| REINFORCE | policy gradient, on-policy | discrete/continuous | Monte Carlo returns |
| A2C/A3C | actor-critic, on-policy | discrete/continuous | learned value baseline |
| PPO | actor-critic, on-policy | discrete/continuous | clipped policy objective |
| DDPG/TD3 | actor-critic, off-policy | continuous | deterministic actor, critic |
| SAC | actor-critic, off-policy | continuous | entropy-regularized policy |


### Environment/Dataset Explanation

RL does not start with a fixed labeled dataset. The agent creates data by interacting with an environment.

Each transition is:

$$ (state, action, reward, next\_state, done) $$

For DQN, this transition goes into a replay buffer. For PPO, recent trajectories are collected using the current policy and then discarded after a few update epochs.


```python
import pandas as pd
trajectory = pd.DataFrame({
    'state':['s0','s1','s2','s3'],
    'action':['right','right','jump','right'],
    'reward':[0,0,1,5],
    'next_state':['s1','s2','s3','terminal'],
    'done':[False,False,False,True]
})
display(trajectory)
```

**Metric meaning:** episode return is the main learning signal, but also inspect episode length, success rate, policy entropy, value loss, and instability across seeds.
