## Research-Backed Coverage Expansion: Modern CNN Architectures and Transfer Learning

This expansion fills concepts commonly covered in deep-learning texts and courses such as the Deep Learning book, CS231n/CS224n, and Dive into Deep Learning. Each concept is written for learning flow: definition, where it is used, why it matters, and common mistakes.


### Architecture Design Pattern

**Definition:** Modern CNNs combine convolution blocks, normalization, activation, downsampling, and residual/efficient connections.

**Where it is used:** CNN architecture reading.

**Why it matters:** Recognizing patterns makes new architectures easier to understand.

**Common mistake:** Memorizing model names without understanding block design.


### AlexNet Contributions

**Definition:** ReLU, GPU training, dropout, data augmentation, and large-scale ImageNet success.

**Where it is used:** CNN history.

**Why it matters:** It showed deep CNNs could dominate visual recognition.

**Common mistake:** Thinking AlexNet matters only because it was bigger.


### VGG Simplicity

**Definition:** Stacks small 3x3 convolutions deeply.

**Where it is used:** Architecture intuition.

**Why it matters:** Several small kernels can replace larger receptive fields with more nonlinearity.

**Common mistake:** Using VGG-style huge dense heads when parameters are limited.


### Inception Multi-Scale Processing

**Definition:** Uses parallel branches with different kernel sizes.

**Where it is used:** GoogLeNet/Inception family.

**Why it matters:** Captures patterns at multiple scales efficiently.

**Common mistake:** Adding branches without managing compute.


### Residual Learning

**Definition:** Learns a residual function F(x) added to x.

**Where it is used:** ResNet and many modern architectures.

**Why it matters:** Improves gradient flow and deep model training.

**Common mistake:** Forgetting shape matching in skip connections.


### Bottleneck Block

**Definition:** Reduces channels, processes, then expands channels.

**Where it is used:** ResNet/Inception/MobileNet efficiency.

**Why it matters:** Cuts computation while keeping representation power.

**Common mistake:** Using bottlenecks too narrow and losing information.


### Depthwise Separable Convolution

**Definition:** Splits spatial filtering and channel mixing.

**Where it is used:** MobileNet and efficient models.

**Why it matters:** Greatly reduces computation.

**Common mistake:** Thinking it always matches full convolution accuracy.


### Compound Scaling

**Definition:** Scales depth, width, and resolution together.

**Where it is used:** EfficientNet.

**Why it matters:** Balanced scaling is better than increasing only one dimension.

**Common mistake:** Increasing image resolution without adjusting model capacity.


### Transfer Learning Modes

**Definition:** Feature extraction freezes backbone; fine-tuning updates some or all layers.

**Where it is used:** Practical vision training.

**Why it matters:** Saves data and compute by reusing pretrained features.

**Common mistake:** Fine-tuning all layers with high LR on tiny data.


### Discriminative Learning Rates

**Definition:** Different learning rates for backbone and head.

**Where it is used:** Fine-tuning pretrained models.

**Why it matters:** Pretrained layers often need smaller updates.

**Common mistake:** Using one high LR for the whole model.


### Domain Shift in Transfer

**Definition:** Pretraining data differs from target data.

**Where it is used:** Medical, satellite, industrial vision.

**Why it matters:** Feature usefulness depends on similarity.

**Common mistake:** Assuming ImageNet transfer always solves domain-specific tasks.


### Model Selection for Deployment

**Definition:** Accuracy, latency, memory, input size, and hardware matter.

**Where it is used:** Mobile/edge/server inference.

**Why it matters:** Best validation accuracy may be unusable in production.

**Common mistake:** Choosing the largest architecture by default.
