````python
A=np.array([[0,1,1,0],[1,0,1,0],[1,1,0,1],[0,0,1,0]]); X=np.array([[1,0],[0,1],[1,1],[0,0]],float); Ahat=A+np.eye(4); D=np.diag(1/Ahat.sum(axis=1)); H=D@Ahat@X; display(pd.DataFrame(X,columns=['f1','f2']).rename_axis('node')); display(pd.DataFrame(H,columns=['msg_f1','msg_f2']).round(3))
````

````python
plt.imshow(A,cmap='Blues'); plt.title('adjacency matrix'); plt.colorbar(); plt.show()
````
