## Research-Backed Coverage Expansion: Graph Neural Networks

This expansion fills concepts commonly covered in deep-learning texts and courses such as the Deep Learning book, CS231n/CS224n, and Dive into Deep Learning. Each concept is written for learning flow: definition, where it is used, why it matters, and common mistakes.


### Homophily

**Definition:** Connected nodes often have similar labels/features.

**Where it is used:** Citation/social graphs.

**Why it matters:** GCNs work well when homophily holds.

**Common mistake:** Assuming every graph is homophilous.


### Heterophily

**Definition:** Connected nodes may be different.

**Where it is used:** Fraud, transaction, biological, some web graphs.

**Why it matters:** Standard GCN smoothing can hurt.

**Common mistake:** Using GCN without checking graph label patterns.


### Self-Loops

**Definition:** Edges from a node to itself.

**Where it is used:** GCN message passing.

**Why it matters:** Preserve a node's own features during aggregation.

**Common mistake:** Aggregating only neighbors and losing self information.


### Degree Normalization

**Definition:** Scaling messages by node degree.

**Where it is used:** GCN and random-walk methods.

**Why it matters:** Prevents high-degree nodes from dominating.

**Common mistake:** Summing raw neighbor features on graphs with varied degree.


### Readout Function

**Definition:** Aggregates node embeddings into a graph embedding.

**Where it is used:** Graph classification.

**Why it matters:** Turns variable-size graphs into fixed-size vectors.

**Common mistake:** Using mean readout when important rare nodes are diluted.


### Edge Features

**Definition:** Attributes on edges such as relation type, distance, weight.

**Where it is used:** Molecular bonds, transactions, knowledge graphs.

**Why it matters:** Edges can carry critical information.

**Common mistake:** Treating all edges as identical.


### Heterogeneous Graph

**Definition:** Graph with multiple node or edge types.

**Where it is used:** Knowledge graphs, recommender systems.

**Why it matters:** Requires type-aware modeling.

**Common mistake:** Forcing all node types into one feature schema.


### Graph Attention Coefficients

**Definition:** Learned neighbor importance weights.

**Where it is used:** GAT models.

**Why it matters:** Not all neighbors should contribute equally.

**Common mistake:** Reading attention as perfect explanation.


### Neighbor Sampling

**Definition:** Sampling local neighborhoods for mini-batch training.

**Where it is used:** Large-scale GNNs.

**Why it matters:** Full graph neighborhoods can be too large.

**Common mistake:** Training only on tiny induced subgraphs without care.


### Inductive vs Transductive

**Definition:** Inductive generalizes to new nodes/graphs; transductive predicts within a fixed graph.

**Where it is used:** GNN evaluation design.

**Why it matters:** Defines valid train/test splits.

**Common mistake:** Leaking test graph structure into training improperly.


### Oversquashing

**Definition:** Too much distant information compressed into fixed-size embeddings.

**Where it is used:** Long-range graph dependencies.

**Why it matters:** Limits ability to use far-away signals.

**Common mistake:** Only addressing oversmoothing and missing oversquashing.


### Graph Leakage

**Definition:** Edges or features reveal labels across splits.

**Where it is used:** Graph ML evaluation.

**Why it matters:** Can create fake high performance.

**Common mistake:** Random node split when connected test labels leak through graph.


```python
import numpy as np, matplotlib.pyplot as plt
A=np.array([[1,1,1,0],[1,1,0,0],[1,0,1,1],[0,0,1,1]], dtype=float) # with self-loops
X=np.array([[1,0],[1,1],[0,1],[0,2]], dtype=float)
D=np.diag(1/np.sqrt(A.sum(axis=1)))
H=D@A@D@X
print('original node features:\n',X)
print('one normalized GCN aggregation step:\n',np.round(H,3))
plt.imshow(H,cmap='viridis'); plt.title('Node features after one GCN-style aggregation'); plt.colorbar();
```

## Algorithm and Dataset Expansion: GNN Algorithms and Graph Datasets

This section connects the concepts to algorithms and datasets. For each module, the learner should understand what data shape is appropriate, which algorithm family fits the data, what metric to read, and what failure pattern to check.


### GNN Algorithm Map

| Algorithm | Aggregation idea | Good for | Limitation |
|---|---|---|---|
| GCN | normalized neighbor averaging | homophilous graphs | oversmoothing/heterophily |
| GraphSAGE | sampled neighborhood aggregation | inductive large graphs | sampling choices matter |
| GAT | attention over neighbors | unequal neighbor importance | attention cost |
| GIN | expressive sum aggregation + MLP | graph classification | can overfit |
| R-GCN | relation-specific edges | knowledge graphs | many relation parameters |
| GAE/VGAE | graph autoencoding | link prediction | negative sampling matters |


### Dataset Explanation

A graph dataset is not rows and columns only. It has:

* nodes;
* edges;
* node features;
* edge features or edge types;
* labels for nodes, edges, or whole graphs.

The split must respect the task. Node classification, graph classification, and link prediction require different splits and metrics.


```python
import numpy as np, pandas as pd
nodes=pd.DataFrame({'node':[0,1,2,3],'feature_age':[21,25,40,42],'label':['A','A','B','B']})
edges=pd.DataFrame({'src':[0,1,2,2],'dst':[1,2,3,0],'edge_type':['friend','follows','friend','follows']})
display(nodes)
display(edges)
print('node classification target shape:', nodes['label'].shape)
print('edge list shape:', edges[['src','dst']].shape)
```
