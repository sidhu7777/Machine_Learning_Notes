````python
x=np.sin(np.linspace(0,2*np.pi,100)); fig,ax=plt.subplots(1,4,figsize=(12,3))
for i,n in enumerate([0,.2,.6,1.]): ax[i].plot(x+n*rng.normal(size=len(x))); ax[i].set_title(f'noise={n}')
plt.show()
````
