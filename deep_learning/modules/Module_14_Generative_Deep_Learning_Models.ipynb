## Research-Backed Coverage Expansion: Generative Deep Learning Models

This expansion fills concepts commonly covered in deep-learning texts and courses such as the Deep Learning book, CS231n/CS224n, and Dive into Deep Learning. Each concept is written for learning flow: definition, where it is used, why it matters, and common mistakes.


### Explicit Density Model

**Definition:** A model that directly defines likelihood p(x).

**Where it is used:** Autoregressive models, normalizing flows.

**Why it matters:** Allows likelihood evaluation.

**Common mistake:** Assuming high likelihood always means good samples.


### Implicit Generative Model

**Definition:** A model that samples without explicit likelihood.

**Where it is used:** GANs.

**Why it matters:** Can produce sharp samples.

**Common mistake:** Harder to evaluate likelihood.


### Autoregressive Factorization

**Definition:** Models data as product of conditional probabilities.

**Where it is used:** Language models, PixelCNN/WaveNet.

**Why it matters:** Turns generation into next-step prediction.

**Common mistake:** Forgetting generation is sequential and can be slow.


### VAE ELBO

**Definition:** Evidence lower bound combines reconstruction and KL terms.

**Where it is used:** Variational autoencoders.

**Why it matters:** Trains approximate latent-variable models.

**Common mistake:** Ignoring KL collapse.


### Posterior Collapse

**Definition:** Decoder ignores latent variable in VAE.

**Where it is used:** Strong decoders and text VAEs.

**Why it matters:** Latent space becomes useless.

**Common mistake:** Only watching reconstruction loss.


### GAN Minimax Game

**Definition:** Generator and discriminator learn against each other.

**Where it is used:** GAN training.

**Why it matters:** Can produce realistic samples but training is unstable.

**Common mistake:** Expecting monotonic loss curves like supervised learning.


### Mode Collapse

**Definition:** Generator produces limited variety.

**Where it is used:** GAN failure mode.

**Why it matters:** Samples look good but diversity is poor.

**Common mistake:** Judging GAN only from a few cherry-picked images.


### Normalizing Flow Invertibility

**Definition:** Uses invertible transformations with tractable Jacobian.

**Where it is used:** Density estimation and sampling.

**Why it matters:** Allows exact likelihood and sampling.

**Common mistake:** Designing layers that are not invertible.


### Diffusion Forward Process

**Definition:** Gradually adds noise to data.

**Where it is used:** Diffusion model training.

**Why it matters:** Creates supervised denoising steps.

**Common mistake:** Thinking the model learns by destroying data only.


### Denoising Score

**Definition:** Model predicts noise or score needed to reverse corruption.

**Where it is used:** Diffusion generation.

**Why it matters:** Sampling reverses noise step by step.

**Common mistake:** Ignoring sampler choice and number of steps.


### Classifier-Free Guidance

**Definition:** Combines conditional and unconditional predictions.

**Where it is used:** Text-to-image diffusion.

**Why it matters:** Controls condition strength.

**Common mistake:** Using guidance too high and reducing diversity/artifacts.


### Generative Evaluation

**Definition:** FID, IS, precision/recall, likelihood, human eval.

**Where it is used:** Generative model comparison.

**Why it matters:** No single metric captures quality and diversity.

**Common mistake:** Trusting one metric blindly.


```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(2)
x=np.sin(np.linspace(0,2*np.pi,200))
noise_levels=[0.05,0.2,0.5,1.0]
for nl in noise_levels:
    plt.plot(x + rng.normal(scale=nl, size=x.shape), alpha=.7, label=f'noise {nl}')
plt.title('Diffusion intuition: data is gradually corrupted by noise')
plt.xlabel('position'); plt.ylabel('value'); plt.legend();
```
