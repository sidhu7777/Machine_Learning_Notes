# Module 09: Reinforcement Learning (RL) Foundations


## Module outline

* **Step 1: The Core Framework**
  * Elements: Agent, Environment, State ($S$), Action ($A$), Reward ($R$), State Transition Probability ($P$), and Discount Factor ($\gamma$).
  * Policies: Deterministic policy vs. Stochastic policy ($\pi(a|s)$).
  * Exploration vs. Exploitation Tradeoff: $\epsilon$-greedy exploration, Upper Confidence Bound (UCB) selection.
* **Step 2: Mathematical Representation & Bellman Equations**
  * Markov Decision Processes (MDP) and the Markov Property.
  * State-Value Function: $V^{\pi}(s) = \mathbb{E}_{\pi} \left[ \sum_{t=0}^{\infty} \gamma^t R_{t+1} \mid S_0 = s \right]$.
  * Action-Value Function: $Q^{\pi}(s, a) = \mathbb{E}_{\pi} \left[ \sum_{t=0}^{\infty} \gamma^t R_{t+1} \mid S_0 = s, A_0 = a \right]$.
  * Bellman Expectation and Optimality Equations.
  * Dynamic Programming: Policy Evaluation, Policy Iteration, and Value Iteration.
* **Step 3: Model-Free RL Algorithms**
  * Monte Carlo Methods: First-visit and Every-visit Monte Carlo prediction.
  * Temporal Difference (TD) Learning: TD(0) prediction, step-by-step updates.
  * Q-Learning: Off-policy temporal difference control ($Q(s, a) \leftarrow Q(s, a) + \alpha [R + \gamma \max_{a'} Q(s', a') - Q(s, a)]$).
  * SARSA (State-Action-Reward-State-Action): On-policy temporal difference control.
